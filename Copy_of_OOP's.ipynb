```python
class palindrome:
  def check_string(self):
    name=input(" Enter the string : ")
    if name==name[::-1]:
      return True,name
    else:
      return False,name
p=palindrome()                #p is variable but palindrome is object
res=p.check_string()
print(res)
```

```python
class Datatype:
  def check_datatype(self):
    data=eval(input("enter the data ...."))
    if isinstance(data,(list,tuple)):
      return True,data
    else:
      return False,data,type(data)

d=Datatype()
res=d.check_datatype()

print(res)
```

```python
# class classname:
#   @classmethod
#   def fun(cls,arg1,arg2,..........)
#   classname.func()
#   (or)
#   obj.func()


class Sample:
  @classmethod
  def add(self):
    print("im a class method")
Sample.add()
s=Sample()
s.add()
```

```python
class Student:
  def __init__(self,name,grade,percentage):
    self.name=name
    self.grade=grade
    self.percentage=percentage
  def student_details(self):
    print(f'{self.name} is in class {self.grade}, with {self.percentage}%')

student1=Student("raj",11,84)
student1.student_details()
print(student1)
```
